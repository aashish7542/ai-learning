Mean Square Error

```python
y_true = [3, 5, 2, 7]
y_pred = [2, 5, 4, 8]

# 1. Define the function
def mse(y_true, y_pred):
  if len(y_pred) != len(y_true):
    raise ValueError("lengths don't match")
  n = len(y_pred)
  if n == 0:
    raise ValueError("empty list")
  return sum((a-b)**2 for a,b in zip(y_true,y_pred)) / n
```

```python
# 2. Test the function
assert mse([1, 2], [1, 2]) == 0
assert mse([1, 2], [2, 1]) == 1
assert mse([1],[1]) == 0

try:
    mse([], [])
    assert False, "Expected ValueError"
except ValueError:
    print("Test passed")


mse(y_true, y_pred)
```

Mean Absolute Error

```python
# 1. Define the function
def mae(y_true, y_pred):
  if len(y_pred) != len(y_true):
    raise ValueError("lengths don't match")
  n = len(y_pred)
  if n == 0:
    raise ValueError("empty list")
  return sum(abs(a-b) for a,b in zip(y_true,y_pred)) / n

# 2. Test the function
assert mae([1, 2], [1, 2]) == 0
assert mae([1, 2], [2, 1]) == 1
assert mae([1],[1]) == 0

try:
    mae([], [])
    assert False, "Expected ValueError"
except ValueError:
    print("Test passed")


mse(y_true, y_pred)
```

Now we will do gradient descent from scratch

```python
import math

# The original function L(x)
def L(x):
    return (x - 5) ** 2

# The derivative function L'(x)
def L_prime(x):
    return 2 * (x - 5)

# Gradient descent; alpha is the step size
def gradient_descent(x, alpha, steps):
    for i in range(steps):
        x = x - alpha * L_prime(x)
    result = gradient_descent(x, alpha, steps)
    print("Final x:", result)

#main function
def main():
    x = input("Starting point: ")
    alpha = input("Step size: ")
    steps = input("Number of steps: ")
    return gradient_descent(float(x), float(alpha), int(steps))


# Test the function
assert math.isclose(gradient_descent(0, 0.1, 100), 5, abs_tol=1e-7)
assert gradient_descent(5, 0.1, 0) == 5

# try:
#     gradient_descent(5, 0.1, 0)
#     assert False, "Expected ValueError"
# except ValueError:
#     print("Test passed")

# gradient_descent(0, 0.1, 100)

main()

```
