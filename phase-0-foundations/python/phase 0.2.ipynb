Now, we will execute the same MSE function that we did earlier, but this time, we will use NumPy arrays

```python
import numpy as np

y_true = np.array([3, 5, 2, 7])
y_pred = np.array([2, 5, 4, 8])

mse = ((y_true - y_pred) ** 2).mean()
print(mse)
mae = np.abs(y_true - y_pred).mean()
print(mae)
```

**Matrix Multiplication**

```python
import numpy as np

A = np.array([
    [2, 7, 1],
    [4, 6, 1],
    [6, 8, 1]])

W = np.array([5, 2, 30])

AW = A@W
print(AW)
```

**Gradient Descent (Vectors)**

```python
import numpy as np

def gradient(A, W, y):
    n = len(y)
    return (2 / n) * A.T @ (A @ W - y)
```

**Final Mini-Project: Training Linear Regression**

```python
import numpy as np

# Training data
X = np.array([1, 2, 3, 4, 5])
y = np.array([50, 60, 70, 80, 90])

# Starting with deliberately incorrect parameters
w = 1.0
b = 0.0

def predict(X, w, b):
    return w * X + b

def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

print("Predictions:", predict(X, w, b))
print("Initial MSE:", mse(y, predict(X, w, b)))
```
